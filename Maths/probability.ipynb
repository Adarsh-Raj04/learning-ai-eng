# Probability

- **Probability**: It is a number between 0 and 1. It tell "How likely is something to happen".

- **Experiment**: Something we perform. Ex: Roll a die, Flip a coin
- **Outcome**: The Result of the experiment. Ex: One possible result like getting head or tell on coin or getting any one number between 1 to 6 on die.
- **Event**: Collection of outcomes that we are interested in. Ex: Getting an even number on die {2,4,6}

---
#### Example:
1. What is the probability of getting an even number on die.
    ```text
    Even Numbers: 2,4,6 that is 3
    Total Possible Outcomes: 1,2,3,4,5,6 that is 6
    P(even number): 3/6 = 0.5   means there is 50% chance of getting an even number on die.
    ```

**Conditional Probability:** - What is probability of A after I already know B.

Suppose:

```text
A = person has a disease
B = medical test is positive
```
Instead of asking:

> What's the probability someone has the disease?

we might ask:

> What's the probability someone has the disease **given that their test is positive**?

That's:

$$
P(A|B)
$$

Read it as:

> Probability of A **given B**.

The formula is:

$$
\boxed{P(A|B)=\frac{P(A\cap B)}{P(B)}}
$$

### Independence
- Two events are independent if knowing one happened doesn't change the probability of the other.

Example:
```text
Flip a coin
Roll a die
```
The coin result doesn't affect die result.
So: 
$$
P(A\cap B)=P(A)P(B)
$$

## Bayes' Theorem

$$
\boxed{
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
}
$$

## Random Variable
- A random variable is basically a variable whose value depends on the outcome of a random process.

Example:
X = result of rolling a die

then 
```text
X ∈ {1,2,3,4,5,6}
```

### Expected Value
- It is essentially the long-run average.
Suppose a fair die is rolled repeatedly.

Possible values:

```text
1, 2, 3, 4, 5, 6
```

Each has probability `1/6`.

Expected value:

$$
E[X]
=
1\frac16+
2\frac16+
3\frac16+
4\frac16+
5\frac16+
6\frac16
$$

which gives:

$$
E[X]=3.5
$$

## Variance
Mean tells us:

> Where are the values centered?

Variance tells us:

> **How spread out are they?**


Imagine two datasets:

```text
A = [49, 50, 51]
B = [10, 50, 90]
```

Both have mean:

```text
50
```

But B is much more spread out.

Therefore:

```text
Variance(A) → small
Variance(B) → large
```

Standard deviation is simply:

$$
\text{Std}=\sqrt{\text{Variance}}
$$
