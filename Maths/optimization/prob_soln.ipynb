## Q1. MSE Loss

Given:

$$
y = [2, 4, 6]
$$

$$
\hat{y} = [1, 5, 7]
$$

The Mean Squared Error is:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

### Step 1 — Calculate the Errors

$$
2-1=1
$$

$$
4-5=-1
$$

$$
6-7=-1
$$

### Step 2 — Square the Errors

$$
1^2=1
$$

$$
(-1)^2=1
$$

$$
(-1)^2=1
$$

### Step 3 — Calculate MSE

$$
MSE = \frac{1+1+1}{3}
$$

$$
\boxed{MSE=1}
$$

## Q2. Gradient Descent

Given:

$$
L(w)=w^2
$$

$$
w=5
$$

$$
\eta=0.1
$$

### Step 1 — Calculate the Gradient

Differentiate the loss function:

$$
\frac{dL}{dw}=2w
$$

At $w=5$:

$$
\frac{dL}{dw}=2(5)=10
$$

Therefore:

$$
\boxed{\frac{dL}{dw}=10}
$$

### Step 2 — Update the Parameter

The Gradient Descent update rule is:

$$
w_{\text{new}}
=
w-\eta\frac{dL}{dw}
$$

Substitute the values:

$$
w_{\text{new}}
=
5-(0.1)(10)
$$

$$
w_{\text{new}}
=
5-1
$$

Therefore:

$$
\boxed{w_{\text{new}}=4}
$$

### Final Answer

$$
\boxed{
\frac{dL}{dw}=10,\qquad
w_{\text{new}}=4
}
$$

## Q3. Batch GD vs SGD vs Mini-Batch GD

Given:

$$
10,000\text{ training examples}
$$

Mini-batch size:

$$
32
$$

### Batch Gradient Descent

One update uses the entire dataset:

```text
10,000 examples
       ↓
   Gradient
       ↓
    Update
```

Therefore, one update processes:

$$
\boxed{10,000\text{ examples}}
$$

### Stochastic Gradient Descent (SGD)

One update uses one training example:

```text
1 example
    ↓
 Gradient
    ↓
  Update
```

Therefore, one update processes:

$$
\boxed{1\text{ example}}
$$

### Mini-Batch Gradient Descent

One update uses 32 training examples:

```text
32 examples
     ↓
  Gradient
     ↓
   Update
```

Therefore, one update processes:

$$
\boxed{32\text{ examples}}
$$

### Summary

| Method | Examples per Update |
|---|---:|
| Batch GD | 10,000 |
| SGD | 1 |
| Mini-Batch GD | 32 |

## Q4. Why Subtract the Gradient?

Gradient Descent uses:

$$
\theta_{\text{new}}
=
\theta-\eta\nabla L
$$

instead of:

$$
\theta_{\text{new}}
=
\theta+\eta\nabla L
$$

because the gradient points toward the direction of **steepest increase** in the loss.

We want to **decrease** the loss.

Therefore, we move in the opposite direction:

$$
\boxed{-\nabla L}
$$

### Example

Suppose:

$$
\nabla L=+10
$$

and:

$$
\eta=0.1
$$

Then:

$$
\theta_{\text{new}}
=
\theta-(0.1)(10)
$$

$$
\theta_{\text{new}}
=
\theta-1
$$

So the parameter moves downward.

If we added the gradient:

$$
\theta_{\text{new}}
=
\theta+(0.1)(10)
$$

the parameter would move in the direction of increasing loss.

### Final Answer

$$
\boxed{
\text{Subtract the gradient because we want to move toward lower loss.}
}
$$

## Q5. Learning Rate Too Large

Suppose:

$$
\eta=10
$$

and the loss keeps jumping wildly instead of decreasing smoothly.

### Answer

The learning rate is likely **too large**.

A very large learning rate can cause the optimizer to take excessively large steps:

```text
        Overshoot
           ↓
Minimum ←──────→ Overshoot
           ↑
        Overshoot
```

Instead of gradually approaching the minimum, the parameters can repeatedly jump across it or cause training to diverge.

A smaller learning rate would generally produce more controlled updates.

### Final Answer

$$
\boxed{
\text{Likely issue: Learning rate is too high.}
}
$$
