# Phase 1 — Optimization Notes

## 1. Objective Function / Loss Function

### Definition

An **objective function** is the function we want to optimize.

In ML, we usually minimize a **loss function**, which measures how wrong the model's predictions are.

### Formula

For Mean Squared Error (MSE):

$$
L = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Where:

- $n$ = number of training examples
- $y_i$ = actual value
- $\hat{y}_i$ = predicted value

### ML Relevance

Training means finding model parameters that minimize the loss:

$$
\boxed{\text{Find parameters that minimize } L}
$$

Typical flow:

```text
Model
  ↓
Prediction
  ↓
Loss
  ↓
Optimization
```

---

## 2. Convexity

### Definition

A function is **convex** if it has a bowl-like shape with a global minimum.

### Formula

A function $f(x)$ is convex if:

$$
f(\lambda x_1+(1-\lambda)x_2)
\leq
\lambda f(x_1)+(1-\lambda)f(x_2)
$$

Where:

- $x_1, x_2$ = two points
- $0 \leq \lambda \leq 1$

The full mathematical definition is less important here than the intuition.

### Intuition

For a convex function, if you find a local minimum, it is also the global minimum.

$$
\boxed{\text{Local Minimum} = \text{Global Minimum}}
$$

### ML Relevance

Convex optimization problems are generally easier to optimize because there are no separate local minima that are better or worse than the global minimum.

Examples:

- Linear regression with MSE → generally convex
- Neural network loss landscapes → generally non-convex

## 3. Gradient Descent

### Definition

**Gradient Descent** is an optimization algorithm that repeatedly changes model parameters in the direction that reduces the loss.

### Formula

$$
\boxed{
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla L(\theta)
}
$$

Where:

- $\theta$ = model parameters
- $L(\theta)$ = loss function
- $\nabla L(\theta)$ = gradient of the loss
- $\eta$ = learning rate

The gradient points in the direction of **steepest increase** of the loss.

Therefore, we move in the **opposite direction** to reduce the loss.

### ML Relevance

Gradient Descent is fundamental to training models such as:

- Linear regression
- Logistic regression
- Neural networks
- Deep learning models

### Core Loop

```text
Initialize Parameters
        ↓
      Predict
        ↓
  Calculate Loss
        ↓
Calculate Gradient
        ↓
Update Parameters
        ↓
      Repeat
```

---

## 4. Stochastic Gradient Descent (SGD)

### Definition

**Stochastic Gradient Descent (SGD)** calculates the gradient using **one training example at a time** instead of the entire dataset.

### Formula

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta\nabla L_i(\theta_t)
}
$$

Where:

- $\theta_t$ = parameters at step $t$
- $\eta$ = learning rate
- $L_i$ = loss for one training example
- $\nabla L_i$ = gradient calculated from that example

### ML Relevance

SGD allows frequent parameter updates and can be practical for very large datasets.

However, because each update uses only one example, the updates are noisier than full-batch Gradient Descent.

### Intuition

```text
Dataset
   ↓
Pick ONE example
   ↓
Calculate Loss
   ↓
Calculate Gradient
   ↓
Update Parameters
   ↓
Repeat
```

---

## 5. Mini-Batch Gradient Descent

### Definition

**Mini-Batch Gradient Descent** calculates the gradient using a small group of training examples.

It sits between:

- **Batch Gradient Descent** → entire dataset
- **SGD** → one example
- **Mini-Batch Gradient Descent** → small batch

### Formula

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta
\frac{1}{B}
\sum_{i=1}^{B}
\nabla L_i(\theta_t)
}
$$

Where:

- $B$ = batch size
- $\eta$ = learning rate
- $L_i$ = loss for example $i$
- $\theta_t$ = parameters at step $t$

### ML Relevance

Mini-batch training is the standard practical approach for training neural networks because it provides:

- Efficient computation
- Better hardware utilization
- More stable updates than single-example SGD
- Faster training than processing the entire dataset for every update

### Comparison

| Method | Data Used Per Update | Gradient Behavior |
|---|---|---|
| Batch GD | Entire dataset | Stable |
| SGD | 1 example | Noisy |
| Mini-Batch GD | Small batch | Less noisy |

# Big Picture

The optimization process can be summarized as:

$$
\boxed{
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Parameter Update}
\rightarrow
\text{Repeat}
}
$$

The five concepts fit together as:

```text
Objective / Loss
        ↓
What are we trying to minimize?

Convexity
        ↓
What does the optimization landscape look like?

Gradient Descent
        ↓
How do we move toward lower loss?

SGD / Mini-Batch
        ↓
How do we calculate gradients efficiently?
```
