# Phase 1 — Probability Notes

## 1. Random Variable

A random variable maps an uncertain outcome to a numerical value.

* **Discrete:** countable values.
* **Continuous:** any value in an interval.

### Formula

$$
P(X=x)
$$

where:

* $X$ is the random variable.
* $x$ is a possible value.

### ML Relevance

Models use quantities such as $P(Y|X)$.

LLMs model:

$$
P(\text{next token} \mid \text{context})
$$

---

## 2. Probability Distribution

A probability distribution describes how probability is assigned across possible values.

For a discrete random variable:

$$
\sum_x P(X=x)=1
$$

### ML Relevance

* Classifiers output probability distributions.
* LLMs produce a probability distribution over possible next tokens.

---

## 3. Conditional Probability

Conditional probability represents the probability of event $A$ given that event $B$ has occurred.

### Formula

$$
\boxed{
P(A|B)=\frac{P(A\cap B)}{P(B)}
}
$$

### ML Relevance

$P(Y|X)$ is fundamental to probabilistic prediction.

LLM next-token prediction can also be expressed as:

$$
P(\text{next token}|\text{context})
$$

---

## 4. Bayes' Theorem

Bayes' theorem allows us to reverse a conditional probability.

### Formula

$$
\boxed{
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
}
$$

Where:

* **Posterior:** $P(A|B)$
* **Likelihood:** $P(B|A)$
* **Prior:** $P(A)$
* **Evidence:** $P(B)$

### ML Relevance

Bayes' theorem is used in:

* Bayesian Machine Learning
* Naive Bayes
* Uncertainty estimation
* Probabilistic inference

---

## 5. Expectation

Expectation is the **probability-weighted average** of a random variable.

### Formula

$$
\boxed{
E[X]=\sum_x xP(X=x)
}
$$

### ML Relevance

Expectation appears in:

* Expected loss
* Expected reward
* Statistical estimation
* Reinforcement Learning

---

## 6. Variance

Variance measures how much values are spread around the mean.

### Formula

$$
\boxed{
Var(X)=E[(X-\mu)^2]
}
$$

where:

$$
\mu=E[X]
$$

An equivalent formula is:

$$
\boxed{
Var(X)=E[X^2]-(E[X])^2
}
$$

### ML Relevance

Variance is important for:

* Understanding uncertainty
* Measuring variability
* Bias-variance analysis
* Statistical modeling

---

## 7. Covariance

Covariance measures how two variables change together.

### Formula

$$
\boxed{
Cov(X,Y)=E[(X-E[X])(Y-E[Y])]
}
$$

### Interpretation

* **Positive covariance:** $X$ and $Y$ tend to move together.
* **Negative covariance:** $X$ and $Y$ tend to move in opposite directions.
* **Near-zero covariance:** little or no linear relationship.

### ML Relevance

Covariance is important in:

* Feature relationships
* Covariance matrices
* Principal Component Analysis (PCA)
* Multivariate statistics

---

## 8. Bernoulli Distribution

The Bernoulli distribution models **one binary experiment**.

The random variable can take only two values:

$$
X\in\{0,1\}
$$

Usually:

$$
P(X=1)=p
$$

and:

$$
P(X=0)=1-p
$$

### Probability Mass Function

$$
\boxed{
P(X=x)=p^x(1-p)^{1-x}
}
$$

where:

$$
x\in\{0,1\}
$$

### ML Relevance

Bernoulli distributions are commonly used for:

* Binary classification
* Logistic regression
* Binary outcomes

---

## 9. Binomial Distribution

The binomial distribution models the number of successes in $n$ independent Bernoulli trials.

### Formula

$$
\boxed{
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
}
$$

Where:

* $n$ = number of trials
* $k$ = number of successes
* $p$ = probability of success

### ML Relevance

Useful for modeling:

* Repeated binary outcomes
* Classification counts
* Number of successful predictions

---

## 10. Gaussian / Normal Distribution

The Gaussian distribution, also called the Normal distribution, is a continuous, bell-shaped probability distribution.

### Notation

$$
\boxed{
X\sim N(\mu,\sigma^2)
}
$$

Where:

* $\mu$ = mean
* $\sigma^2$ = variance
* $\sigma$ = standard deviation

### ML Relevance

Gaussian distributions are commonly used for:

* Noise modeling
* Regression
* Probabilistic Machine Learning
* Bayesian models
* Diffusion-related concepts

---

## 11. Likelihood

Likelihood measures how well a particular parameter value explains the observed data.

Given parameters $\theta$ and observed data:

$$
\boxed{
L(\theta|Data)=P(Data|\theta)
}
$$

Here:

* $Data$ = observed data
* $\theta$ = model parameters

### Important Distinction

Probability asks:

> Given the parameters, how likely is the data?

Likelihood asks:

> Given the observed data, which parameter values make the data more plausible?

### ML Relevance

Likelihood is the foundation of:

* Maximum Likelihood Estimation (MLE)
* Many statistical estimation methods
* Several Machine Learning objectives

---

## 12. Maximum Likelihood Estimation (MLE)

Maximum Likelihood Estimation chooses the parameter values that maximize the likelihood of the observed data.

### Formula

$$
\boxed{
\theta_{MLE}
=
\arg\max_{\theta} P(Data|\theta)
}
$$

Usually, we maximize the **log-likelihood** instead.

$$
\boxed{
\theta_{MLE}
=
\arg\max_{\theta}
\log P(Data|\theta)
}
$$

### Why Use Log-Likelihood?

Suppose the likelihood contains a product:

$$
P(Data|\theta)
=
P(x_1|\theta)P(x_2|\theta)\cdots P(x_n|\theta)
$$

Taking the logarithm converts the product into a sum:

$$
\log(ab)=\log a+\log b
$$

Therefore:

$$
\log\left(
\prod_{i=1}^{n}P(x_i|\theta)
\right)
=
\sum_{i=1}^{n}
\log P(x_i|\theta)
$$

This makes mathematical manipulation and numerical computation much easier.

### Example

Suppose we observe:

$$
H,H,T,H
$$

and:

$$
p=P(H)
$$

The likelihood is:

$$
L(p)=p^3(1-p)
$$

The MLE is:

$$
\hat p_{MLE}
=
\frac{\text{Number of Heads}}{\text{Total Tosses}}
$$

Therefore:

$$
\hat p_{MLE}
=
\frac{3}{4}
=
0.75
$$

### ML Relevance

MLE connects directly to objectives such as:

* Negative Log-Likelihood (NLL)
* Cross-Entropy Loss
* Classification model training

---

## 13. Maximum A Posteriori (MAP)

Maximum A Posteriori estimation chooses parameters using both:

1. The likelihood of the observed data.
2. A prior belief about the parameters.

Bayes' theorem gives:

$$
P(\theta|Data)
\propto
P(Data|\theta)P(\theta)
$$

Therefore:

$$
\boxed{
\theta_{MAP}
=
\arg\max_{\theta}
P(Data|\theta)P(\theta)
}
$$

Where:

* $P(\theta|Data)$ = posterior
* $P(Data|\theta)$ = likelihood
* $P(\theta)$ = prior

### ML Relevance

MAP provides intuition for:

* Bayesian Machine Learning
* Parameter estimation
* Regularization

---

## 14. LLM Token Probabilities

An LLM produces a probability distribution over possible next tokens.

For example:

```text
AI       → 0.60
ML       → 0.25
Python   → 0.15
```

Suppose the context is:

```text
I am learning
```

The model might assign:

$$
P(\text{"AI"}|\text{"I am learning"})=0.60
$$

This means:

> Given the context `"I am learning"`, the model assigns a probability of 60% to `"AI"` as the next token.

The model produces probabilities for many possible tokens, not just the three shown above.
