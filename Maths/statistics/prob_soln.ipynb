## Q1. Mean

Given:

$[10, 20, 30, 40, 50]$

Calculate the mean:

$$
\bar{x} = \frac{10 + 20 + 30 + 40 + 50}{5}
$$

$$
= \frac{150}{5}
$$

$$
\boxed{\bar{x} = 30}
$$

$[10, 12, 13, 15, 100]$

\boxed{\text{Median} = 13}

$$
\boxed{100}

## Q3. Population Variance

Given:

$[2, 4, 6]$

### Step 1: Calculate the Mean

$$
\mu = \frac{2 + 4 + 6}{3}
$$

$$
\mu = 4
$$

### Step 2: Calculate Population Variance

$$
\sigma^2 =
\frac{(2-4)^2 + (4-4)^2 + (6-4)^2}{3}
$$

$$
= \frac{4 + 0 + 4}{3}
$$

$$
= \frac{8}{3}
$$

$$
\boxed{\sigma^2 \approx 2.667}
$$

---

## Q4. Standard Deviation

From Q3:

$$
\sigma^2 = \frac{8}{3}
$$

Standard deviation is the square root of variance:

$$
\sigma = \sqrt{\sigma^2}
$$

$$
\sigma = \sqrt{\frac{8}{3}}
$$

$$
\boxed{\sigma \approx 1.633}
$$

## Q6. Standardization

Given:

$$
x = 80,\qquad \mu = 70,\qquad \sigma = 5
$$

The standardization formula is:

$$
z = \frac{x-\mu}{\sigma}
$$

Substitute the values:

$$
z = \frac{80-70}{5}
$$

$$
z = \frac{10}{5}
$$

$$
\boxed{z = 2}
$$

The value is 2 standard deviations above the mean.

## Q7. 95% Confidence Interval

Given:

$$
\bar{x} = 100,\qquad \sigma = 20,\qquad n = 100
$$

For a 95% normal-based confidence interval:

$$
\bar{x} \pm 1.96\frac{\sigma}{\sqrt{n}}
$$

Substitute the values:

$$
100 \pm 1.96\frac{20}{\sqrt{100}}
$$

$$
= 100 \pm 1.96\frac{20}{10}
$$

$$
= 100 \pm 3.92
$$

Therefore:

$$
\boxed{(96.08,\ 103.92)}
$$

## Q8. Hypothesis Testing

Given:

$$
p = 0.03,\qquad \alpha = 0.05
$$

Compare the p-value with the significance level:

$$
0.03 < 0.05
$$

Therefore:

$$
\boxed{\text{Reject } H_0}
$$

The result is statistically significant at the 5% significance level under the assumptions of the test.

> **Important:** A p-value of 0.03 does not mean that the alternative hypothesis has a 97% probability of being true.

## Q12. RAG Evaluation

Suppose:

- System A = 85%
- System B = 87%

The difference is:

$$
87\% - 85\% = 2\text{ percentage points}
$$

Therefore:

$$
\boxed{\text{Difference} = 2\text{ percentage points}}
$$

However, a point estimate alone is not sufficient to conclude that the difference is statistically meaningful.

We should check:

1. Sample size
2. Uncertainty
3. Confidence intervals
4. Statistical testing
5. Whether both systems were evaluated on the same examples
6. Practical significance

### Conceptual Flow

```text
Evaluation Dataset
        ↓
      Metric
        ↓
     Estimate
        ↓
   Uncertainty
        ↓
Statistical Comparison
        ↓
Practical Significance
        ↓
Engineering Decision
```
