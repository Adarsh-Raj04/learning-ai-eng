# Q1. Functions

Given:

$$
f(x)=3x^2+2x-5
$$

## Find $f(2)$

Substitute $x=2$:

$$
f(2)=3(2)^2+2(2)-5
$$

$$
=3(4)+4-5
$$

$$
=12+4-5
$$

$$
\boxed{f(2)=11}
$$

## Find $f(-1)$

Substitute $x=-1$:

$$
f(-1)=3(-1)^2+2(-1)-5
$$

$$
=3(1)-2-5
$$

$$
=3-2-5
$$

$$
\boxed{f(-1)=-4}
$$

# Q5. Derivative

Given:

$$
f(x)=4x^3+5x^2-7x+10
$$

Use the power rule:

$$
\frac{d}{dx}(x^n)=nx^{n-1}
$$

Differentiate each term:

$$
\frac{d}{dx}(4x^3)=12x^2
$$

$$
\frac{d}{dx}(5x^2)=10x
$$

$$
\frac{d}{dx}(-7x)=-7
$$

$$
\frac{d}{dx}(10)=0
$$

Therefore:

$$
\boxed{f'(x)=12x^2+10x-7}
$$

# Q6. Derivative and Evaluation

Given:

$$
L(w)=w^2+4w+7
$$

Differentiate:

$$
\frac{dL}{dw}=2w+4
$$

Therefore:

$$
\boxed{L'(w)=2w+4}
$$

At $w=3$:

$$
L'(3)=2(3)+4
$$

$$
=6+4
$$

$$
\boxed{L'(3)=10}
$$

> **Note:** $L'(3)$ is the derivative at $w=3$, not the value of the loss $L(3)$.

# Q7. Partial Derivative with Respect to x

Given:

$$
L(x,y)=x^2+3xy+y^2
$$

Find:

$$
\frac{\partial L}{\partial x}
$$

Treat $y$ as a constant.

Differentiate term by term:

$$
\frac{\partial}{\partial x}(x^2)=2x
$$

$$
\frac{\partial}{\partial x}(3xy)=3y
$$

$$
\frac{\partial}{\partial x}(y^2)=0
$$

Therefore:

$$
\boxed{\frac{\partial L}{\partial x}=2x+3y}
$$

---

# Q8. Partial Derivative with Respect to y

Given:

$$
L(x,y)=x^2+3xy+y^2
$$

Find:

$$
\frac{\partial L}{\partial y}
$$

Treat $x$ as a constant.

Differentiate:

$$
\frac{\partial}{\partial y}(x^2)=0
$$

$$
\frac{\partial}{\partial y}(3xy)=3x
$$

$$
\frac{\partial}{\partial y}(y^2)=2y
$$

Therefore:

$$
\boxed{\frac{\partial L}{\partial y}=3x+2y}
$$

# Q9. Gradient

Given:

$$
L(x,y)=x^2+2y^2
$$

The gradient is:

$$
\nabla L=
\begin{bmatrix}
\frac{\partial L}{\partial x}\\
\frac{\partial L}{\partial y}
\end{bmatrix}
$$

First:

$$
\frac{\partial L}{\partial x}=2x
$$

Second:

$$
\frac{\partial L}{\partial y}=4y
$$

Therefore:

$$
\nabla L=
\begin{bmatrix}
2x\\
4y
\end{bmatrix}
$$

At:

$$
(x,y)=(2,3)
$$

we get:

$$
\nabla L=
\begin{bmatrix}
2(2)\\
4(3)
\end{bmatrix}
=
\begin{bmatrix}
4\\
12
\end{bmatrix}
$$

Therefore:

$$
\boxed{
\nabla L(2,3)=
\begin{bmatrix}
4\\
12
\end{bmatrix}
}
$$

# Q10. Chain Rule

Given:

$$
y=u^2
$$

and:

$$
u=3x+1
$$

We want:

$$
\frac{dy}{dx}
$$

Using the chain rule:

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
$$

First:

$$
\frac{dy}{du}=2u
$$

Second:

$$
\frac{du}{dx}=3
$$

Therefore:

$$
\frac{dy}{dx}=2u(3)
$$

$$
=6u
$$

Since:

$$
u=3x+1
$$

we get:

$$
\boxed{\frac{dy}{dx}=6(3x+1)}
$$

Equivalent form:

$$
\boxed{\frac{dy}{dx}=18x+6}
$$

# Q11. Chain Rule — ML Style

Given:

$$
L=(wx-y)^2
$$

We want:

$$
\frac{\partial L}{\partial w}
$$

Introduce an intermediate variable:

$$
u=wx-y
$$

Then:

$$
L=u^2
$$

Using the chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial u}
\frac{\partial u}{\partial w}
$$

First:

$$
\frac{\partial L}{\partial u}=2u
$$

Second:

$$
\frac{\partial u}{\partial w}=x
$$

Therefore:

$$
\frac{\partial L}{\partial w}=2ux
$$

Substitute:

$$
u=wx-y
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}=2x(wx-y)
}
$$

This is a simple example of how the chain rule lets us calculate how a model parameter affects the loss.

# Q13. Backpropagation

Using:

$$
z=wx+b
$$

$$
L=z^2
$$

Given:

$$
w=2,\quad x=3,\quad b=1
$$

Find:

$$
\frac{\partial L}{\partial w}
$$

Using the chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial z}
\frac{\partial z}{\partial w}
$$

First:

$$
L=z^2
$$

Therefore:

$$
\frac{\partial L}{\partial z}=2z
$$

Second:

$$
z=wx+b
$$

Therefore:

$$
\frac{\partial z}{\partial w}=x
$$

So:

$$
\frac{\partial L}{\partial w}=2zx
$$

From Q12:

$$
z=7
$$

and:

$$
x=3
$$

Therefore:

$$
\frac{\partial L}{\partial w}=2(7)(3)
$$

$$
\boxed{\frac{\partial L}{\partial w}=42}
$$

This is the essence of backpropagation: applying the chain rule backward through the computational graph.

# Q14. Gradient Descent — First Update

Given:

$$
L(w)=w^2
$$

Therefore:

$$
\frac{dL}{dw}=2w
$$

Initial value:

$$
w=5
$$

Learning rate:

$$
\eta=0.1
$$

## Step 1: Calculate the gradient

$$
\frac{dL}{dw}=2w
$$

Substitute $w=5$:

$$
\frac{dL}{dw}=2(5)=10
$$

Therefore, the gradient is:

$$
\boxed{\frac{dL}{dw}=10}
$$

## Step 2: Apply the gradient descent update

The gradient descent formula is:

$$
w_{\text{new}}
=
w-\eta\frac{dL}{dw}
$$

Substitute the values:

$$
w_{\text{new}}=5-(0.1)(10)
$$

$$
=5-1
$$

Therefore:

$$
\boxed{w_{\text{new}}=4}
$$

We now use:

Therefore:

w_{\text{new}}=4-(0.1)(8)

# Q16. Jacobian

Given:

$$
f(x,y)=
\begin{bmatrix}
x^2+y\\
xy
\end{bmatrix}
$$

Let:

$$
f_1=x^2+y
$$

and:

$$
f_2=xy
$$

The Jacobian is:

$$
J=
\begin{bmatrix}
\frac{\partial f_1}{\partial x} &
\frac{\partial f_1}{\partial y}\\
\frac{\partial f_2}{\partial x} &
\frac{\partial f_2}{\partial y}
\end{bmatrix}
$$

Calculate each derivative:

$$
\frac{\partial f_1}{\partial x}=2x
$$

$$
\frac{\partial f_1}{\partial y}=1
$$

$$
\frac{\partial f_2}{\partial x}=y
$$

$$
\frac{\partial f_2}{\partial y}=x
$$

Therefore:

$$
\boxed{
J=
\begin{bmatrix}
2x & 1\\
y & x
\end{bmatrix}
}
$$

# Q17. Hessian

Given:

$$
L(x,y)=x^2+3xy+2y^2
$$

First calculate the first partial derivatives.

## First derivative with respect to $x$

$$
\frac{\partial L}{\partial x}
=
2x+3y
$$

## First derivative with respect to $y$

$$
\frac{\partial L}{\partial y}
=
3x+4y
$$

Now calculate the second derivatives.

## Second derivative with respect to $x$

$$
\frac{\partial^2L}{\partial x^2}=2
$$

## Mixed derivative

$$
\frac{\partial^2L}{\partial x\partial y}=3
$$

## Other mixed derivative

$$
\frac{\partial^2L}{\partial y\partial x}=3
$$

## Second derivative with respect to $y$

$$
\frac{\partial^2L}{\partial y^2}=4
$$

Therefore the Hessian is:

$$
\boxed{
H=
\begin{bmatrix}
2 & 3\\
3 & 4
\end{bmatrix}
}
$$

# Final Calculus Summary

The important flow for AI Engineering:

```text
Function
   ↓
Derivative
   ↓
Partial Derivatives
   ↓
Gradient
   ↓
Chain Rule
   ↓
Computational Graph
   ↓
Backpropagation
   ↓
Gradient Descent
   ↓
Neural Network Training
```

## What to remember

- **Derivative** → how much output changes when an input changes.
- **Gradient** → how loss changes with respect to all parameters.
- **Chain Rule** → how changes propagate through multiple operations.
- **Backpropagation** → chain rule applied backward through a neural network.
- **Gradient Descent** → use those gradients to update weights and reduce loss.
- **Jacobian** → derivatives of multiple outputs with respect to multiple inputs.
- **Hessian** → second-order derivatives describing curvature of the loss surface.
