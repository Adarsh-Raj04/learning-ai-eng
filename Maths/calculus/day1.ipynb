
### Calculus tells an AI model how its output changes when we change its parameters.

| Topic | Short definition | Formula | ML relevance |
|---|---|---|---|
| **Function** | Maps an input to an output. | $y=f(x)$ — $x$ = input, $y$ = output | A model is essentially a function: $\hat{y}=f(x;\theta)$. |
| **Limit** | The value a function approaches as input approaches a point. | $\lim_{x\to a} f(x)$ — $a$ = target input | Foundation for derivatives; helps understand continuous model behavior. |
| **Derivative** | Measures how fast output changes when input changes; slope at a point. | $\frac{dy}{dx}$ — $x$ = input, $y$ = output | Tells us how changing a parameter affects model loss. |
| **Partial Derivative** | Derivative of a function with respect to one variable while keeping others fixed. | $\frac{\partial L}{\partial w_i}$ — $L$ = loss, $w_i$ = one parameter | Models have many parameters, so we need each parameter's contribution to loss. |
| **Gradient** | Collection of all partial derivatives; points toward steepest increase. | $\nabla L=\left[\frac{\partial L}{\partial w_1},\ldots,\frac{\partial L}{\partial w_n}\right]$ | Used to determine how model weights should change during training. |
| **Chain Rule** | Calculates how a change propagates through multiple functions. | $\frac{dy}{dx}=\frac{dy}{du}\frac{du}{dx}$ — $u$ = intermediate variable | Foundation of **backpropagation** in neural networks. |
| **Computational Graph** | Represents computations as connected operations. | $x\rightarrow z=wx+b\rightarrow L$ | Allows gradients to be calculated backward through a neural network. |
| **Backpropagation** | Uses the chain rule to calculate gradients from output/loss back toward model parameters. | $\frac{\partial L}{\partial w}=\frac{\partial L}{\partial y}\frac{\partial y}{\partial w}$ | How neural networks determine how each weight contributed to the error. |
| **Gradient Descent** | Iteratively changes parameters in the direction that reduces loss. | $w_{\text{new}}=w_{\text{old}}-\eta\nabla L$ — $w$ = parameter, $\eta$ = learning rate, $\nabla L$ = gradient | Fundamental mechanism for training ML/DL models. |
| **Jacobian** | Matrix containing derivatives of multiple outputs with respect to multiple inputs. | $J_{ij}=\frac{\partial y_i}{\partial x_j}$ — $y_i$ = output, $x_j$ = input | Useful for understanding transformations and gradient flow in complex neural networks. |
| **Hessian** | Matrix of second-order derivatives; describes how the gradient itself changes. | $H_{ij}=\frac{\partial^2L}{\partial w_i\partial w_j}$ | Gives information about the curvature of the loss surface; useful in advanced optimization. |
