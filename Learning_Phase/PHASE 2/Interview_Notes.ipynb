# 14. Interview Takeaways
---

Remember these:

### Q: What is supervised learning?

> Learning a mapping from input features to known target labels using labeled training data.
---
### Q: Regression vs classification?

> Regression predicts a continuous target; classification predicts a discrete class.
---
### Q: What are features?

> Input variables used by the model to make predictions.
---
### Q: What is a label?

> The target/output the model is trained to predict in supervised learning.
---
### Q: Why don't we evaluate on training data?

> Because the model has already seen that data, so the performance may not represent generalization to unseen data.
---
### Q: Why do we need a test set?

> To obtain a final estimate of performance on unseen data after model selection/tuning.
---

### Q: Is Polynomial Regression a linear or nonlinear model?

The nuanced answer:

> It models a nonlinear relationship between the original input and target, but it is linear with respect to its learned coefficients.

That's a great interview answer.

---

### Q: Why can Polynomial Regression overfit?

> Increasing the polynomial degree increases model flexibility, allowing the model to fit noise and training-specific patterns rather than the underlying relationship.

---

### Q: What does increasing `degree` do?

> It creates higher-order polynomial features, increasing model complexity and its ability to represent nonlinear relationships.

---

### Q: Why can Polynomial Regression become expensive?

> Polynomial expansion can dramatically increase the number of features, especially with many original features and interaction terms.

---

### Q: How do you choose the polynomial degree?

> Treat degree as a hyperparameter and evaluate candidate values using validation data or cross-validation, rather than choosing based solely on training error.

---

### Q: Why does Ridge help prevent overfitting?

> It adds an L2 penalty that discourages large coefficients, reducing model complexity and variance.
---
### Q: Does Ridge perform feature selection?

> Not in the strict sense. It shrinks coefficients toward zero but generally does not make them exactly zero.
---
### Q: What happens when alpha increases?

> Regularization becomes stronger, coefficients shrink more, and excessive alpha can cause underfitting.
---
### Q: Why scale features before Ridge?

> Because the regularization penalty operates on coefficient magnitudes, and features with very different scales can make that penalty uneven or distort the effect of regularization.
---
### Q: Ridge vs Linear Regression?

> Ridge adds L2 regularization; ordinary Linear Regression does not.
---
### Q: When would you prefer Ridge?

> When you have many features, correlated features, or concern about overfitting and want to retain all features while shrinking their coefficients.
---

### Q: What is Lasso?

> Lasso is Linear Regression with L1 regularization. It adds the absolute value of coefficients to the loss function, which encourages sparse coefficients and can drive some coefficients exactly to zero.

### Q: Ridge vs Lasso?

> Ridge uses L2 regularization and primarily shrinks coefficients, while Lasso uses L1 regularization and can shrink coefficients exactly to zero, making it useful for feature selection.

### Q: What happens when alpha increases?

> Regularization becomes stronger, so coefficients are pushed more aggressively toward zero. If alpha becomes too large, the model can underfit.

### Q: Does Lasso always remove irrelevant features?

> No. It can set coefficients to zero, but whether a feature is removed depends on the data, regularization strength, scaling, and relationships between features.
