# Machine Learning

- ML is a new way of building systems that learn patterns from data instead of being explicitly programmed with every rule.

```text
Data + Expected Outputs
        ↓
    ML Algorithm
        ↓
      Model
        ↓
  Predictions on new data
```

**Three Major Types of Machine Learning**

At a high level:

```text
                 Machine Learning
                       │
        ┌──────────────┼──────────────┐
        ↓              ↓              ↓
   Supervised     Unsupervised   Reinforcement
        │              │              │
    Has labels      No labels      Reward signal
```
