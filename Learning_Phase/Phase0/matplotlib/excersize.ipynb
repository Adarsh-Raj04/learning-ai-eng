```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]
```

```python
# line plt
plt.plot(x,y)

plt.title("Growth")
plt.xlabel("Days")
plt.ylabel("Value")

plt.show()
```

```python
# Scatter plot
age = [22, 25, 28, 30, 35, 40]
salary = [35000, 45000, 52000, 60000, 75000, 90000]

plt.scatter(age, salary)
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

```python
# histogram of salary distribution
plt.hist(salary, bins=5, color='blue', edgecolor='black')

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

```python
# Barchart
departments = ["AI", "Data", "Engineering"]
employees = [20, 35, 50]

plt.bar(departments, employees, color='green')
plt.title("Number of Employees in Departments")
plt.xlabel("Departments")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# create scatter plot of age vs salary
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Bob"],
    "age": [25, 32, 28, np.nan, 35, 32],
    "salary": [50000, 75000, 62000, 90000, np.nan, 75000],
})


plt.scatter(df["age"], df["salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```
