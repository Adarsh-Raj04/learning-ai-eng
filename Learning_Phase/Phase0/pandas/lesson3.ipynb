```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "age": [25, 32, np.nan, 40, 35, 29],
    "salary": [50000, 75000, 62000, np.nan, 85000, 58000],
    "department": ["AI", "Data", "AI", "Data", "AI", "Data"]
})
```

```python
# check duplicates
print(df.duplicated())

# count the number of duplicates
print(f"Number of duplicates: {df.duplicated().sum()}")

#check duplicates in specific columns
print(df.duplicated(subset=['name', 'age']))

# drop duplicates
df = df.drop_duplicates()
```

```python
# Renaming of columns
df = df.rename(columns={'name': 'employee_name', 'age': 'employee_age', 'salary': 'employee_salary', 'department': 'employee_department'})
print(df.head())  # display the first few rows of the dataframe after renaming columns
```

```python
#change data type
#If a column contains NaN, directly doing:
# df["age"] = df["age"].astype(int) 
# can fail.

# handle missing values first, then convert to int
df["age"] = df["employee_age"].fillna(0).astype(int)  #

print(df)

```

```python
# String Operations

print(df["employee_name"].str.upper()) # convert the employee_name column to uppercase
print(df["employee_name"].str.lower()) # convert the employee_name column to lowercase
print(df["employee_department"].str.contains("AI")) # check if the employee_department column contains the string "AI"
print(df["employee_name"].str.strip()) # remove leading and trailing whitespace from the employee_name column
print(df["employee_department"].str.replace("AI", "Artificial Intelligence")) # replace the string "AI" with "Artificial Intelligence" in the employee_department column
```

```python
df["joining_date"] = [
    "2024-01-15",
    "2023-06-10",
    "2025-02-20",
    "2024-08-15",
    "2023-11-05",
    "2024-03-30"
]
```

```python
df["joining_date"] = pd.to_datetime(df["joining_date"])
```

```python
display(df)
```

```python
# Extract year, month, day from the joining_date column
print(df["joining_date"].dt.year)
print(df["joining_date"].dt.month)
print(df["joining_date"].dt.day)
```
