```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Bob"],
    "age": [25, 32, 28, np.nan, 35, 32],
    "salary": [50000, 75000, 62000, 90000, np.nan, 75000],
    "department": ["AI", "Data", "AI", "Data", "AI", "Data"],
    "joining_date": [
        "2024-01-10",
        "2023-05-20",
        "2024-03-15",
        "2022-08-01",
        "2025-01-10",
        "2023-05-20"
    ]
})
```

```python
#Q1: Find duplicate rows and remove them 
print(f"Duplicate rows:\n{df[df.duplicated()]}")
df = df[~df.duplicated()]



#Q2: Fill missing, in age with meadian and salary with mean
age_median = df["age"].median()
salary_mean = df["salary"].mean()
df["age"].fillna(age_median, inplace=True)
df["salary"].fillna(salary_mean, inplace=True)



#Q3: convert joining_date to datetime and then create joining_year column
df["joining_date"] = pd.to_datetime(df["joining_date"])
df["joining_year"] = df["joining_date"].dt.year


#Q4: Create salary_lakhs column by converting salary to lakhs
df["salary_lakhs"] = df["salary"] / 100000


#Q5 create salary_category column based on salary_lakhs
df["salary_category"] = df["salary"].apply(lambda x: "High" if x>=80000 else "Medium" if x>=60000 else "Low")


#Q6: find the average salary of each department
print(f"Average salary by department:\n{df.groupby('department')['salary'].mean()}")


#Q7: create ML dataset X,Y where X is age and salary and Y is salary_category
X = df[["age", "salary"]].to_numpy()
Y = df["salary_category"].to_numpy()

#Q8: One hot encode on department
pd.get_dummies(df["department"], prefix="dept", drop_first=True)
```
