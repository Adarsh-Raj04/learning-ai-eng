```python
# Dataset Creation

import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

print(X)
print(y)
```

```python
# model

w = 0.0
b = 0.0

y_pred = w * X + b

print(y_pred)
```

```python
# calculate the loss
error = y - y_pred

loss = np.mean(error ** 2)

print("Predictions:", y_pred)
print("Error:", error)
print("Loss:", loss)
```

```python
# calculate gradients
n = len(X)

dw = (-2 / n) * np.sum(X * (y - y_pred))
db = (-2 / n) * np.sum(y - y_pred)

print("dw:", dw)
print("db:", db)
```

```python
learning_rate = 0.01

w = w - learning_rate * dw
b = b - learning_rate * db

print("New w:", w)
print("New b:", b)
```

```python
# new predictions after one update
y_pred = w * X + b

print(y_pred)
```

```python
# calculate loss again
loss = np.mean((y - y_pred) ** 2)

print("New loss:", loss)
```

```python
# automate the learning
import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

for epoch in range(epochs):

    # Forward pass
    y_pred = w * X + b

    # Loss
    loss = np.mean((y - y_pred) ** 2)

    # Gradients
    n = len(X)

    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Update
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 100 == 0:
        print(
            f"Epoch {epoch}: "
            f"Loss={loss:.4f}, "
            f"w={w:.4f}, "
            f"b={b:.4f}"
        )
```

```python
# predictions after training
x_new = 10

prediction = w * x_new + b

print("Prediction:", prediction)
```

```python
# Complete code 
import numpy as np

# Dataset
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
], dtype=float)

y = np.array([
    [3],
    [5],
    [7],
    [9],
    [11]
], dtype=float)

# Parameters
W = np.zeros((1, 1))
b = 0.0

learning_rate = 0.01
epochs = 1000

n = X.shape[0]

for epoch in range(epochs):

    # Forward pass
    y_pred = X @ W + b

    # Error
    error = y - y_pred

    # Loss
    loss = np.mean(error ** 2)

    # Gradients
    dW = (-2 / n) * X.T @ error
    db = (-2 / n) * np.sum(error)

    # Update
    W = W - learning_rate * dW
    b = b - learning_rate * db

    if epoch % 100 == 0:
        print(
            f"Epoch {epoch}: "
            f"Loss={loss:.4f}, "
            f"W={W.ravel()[0]:.4f}, "
            f"b={b:.4f}"
        )
```

```python
# Print the latest weights and bias
print("Final W:", W)
print("Final b:", b)
```

```python
# prediction on new data
X_new = np.array([
    [6],
    [10],
    [20]
], dtype=float)

predictions = X_new @ W + b

print(predictions)
```
