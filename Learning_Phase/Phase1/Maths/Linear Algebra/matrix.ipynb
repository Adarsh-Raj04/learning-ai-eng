# Matrix Addition

Matrices can be added **only when their dimensions match**.

$$
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
$$

$$
B=
\begin{bmatrix}
5&6\\
7&8
\end{bmatrix}
$$

Then:

$$
A+B=
\begin{bmatrix}
1+5&2+6\\
3+7&4+8
\end{bmatrix}
=
\begin{bmatrix}
6&8\\
10&12
\end{bmatrix}
$$

Element-by-element, just like vector addition.

# Matrix Transpose

Transpose essentially **swaps rows and columns**.

$$
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
$$

Then:

$$
A^T=
\begin{bmatrix}
1&4\\
2&5\\
3&6
\end{bmatrix}
$$

Therefore:

$$
(2\times3)\rightarrow(3\times2)
$$

In NumPy:

```python
A.T
```

# 6. Eigenvalues & Eigenvectors

This sounds scary, but the core idea is simple.

A matrix transforms a vector:

$$
Av
$$

Usually, the vector changes **both direction and magnitude**.

But some special vectors only get **scaled**:

$$
Av=\lambda v
$$

Where:

* \(v\) = **eigenvector**
* \(lambda\) = **eigenvalue**

### Example

$$
A=
\begin{bmatrix}
2&0\\
0&3
\end{bmatrix}
$$

Take:

$$
v=
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

Then:

$$
Av=
\begin{bmatrix}
2\\
0
\end{bmatrix}
=
2v
$$

So:

* \(v=[1,0]\) is an eigenvector
* \(lambda=2\) is its eigenvalue

Another eigenvector:

$$
v=
\begin{bmatrix}
0\\1
\end{bmatrix}
$$

gives:

$$
Av=
\begin{bmatrix}
0\\3
\end{bmatrix}
=3v
$$

So eigenvalue = 3.
