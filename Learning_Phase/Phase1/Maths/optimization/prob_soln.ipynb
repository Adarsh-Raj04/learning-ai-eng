## Q1. Gradient Descent

Given:

$$
L(w)=w^2
$$

$$
w=4,\qquad \eta=0.1
$$

### Step 1 — Calculate the Gradient

$$
\frac{dL}{dw}=2w
$$

At $w=4$:

$$
\frac{dL}{dw}=2(4)=8
$$

Therefore:

$$
\boxed{\frac{dL}{dw}=8}
$$

### Step 2 — Update the Parameter

The Gradient Descent update rule is:

$$
w_{\text{new}}
=
w-\eta\frac{dL}{dw}
$$

Substitute the values:

$$
w_{\text{new}}
=
4-(0.1)(8)
$$

$$
w_{\text{new}}
=
4-0.8
$$

Therefore:

$$
\boxed{w_{\text{new}}=3.2}
$$

## Q2. Momentum

### Question

What problem does Momentum primarily help with?

### Solution

Momentum uses information from previous gradients when determining the next update.

It primarily helps by:

- Reducing zig-zagging during optimization
- Smoothing noisy updates
- Accelerating movement in consistent directions

### Final Answer

$$
\boxed{
\text{Momentum makes optimization smoother and can accelerate convergence.}
}
$$

## Q3. SGD vs Mini-Batch GD

### Question

What is the key difference between SGD and Mini-Batch Gradient Descent?

### Solution

### Stochastic Gradient Descent (SGD)

SGD uses **one training example** to calculate the gradient for each update.

$$
\boxed{B=1}
$$

### Mini-Batch Gradient Descent

Mini-Batch GD uses a **small group of training examples**.

$$
\boxed{1<B<n}
$$

where:

- $B$ = batch size
- $n$ = total number of training examples

### Example

Suppose the dataset contains 10,000 examples:

- **SGD** → 1 example per update
- **Mini-Batch GD** → e.g. 32 examples per update

### Final Answer

$$
\boxed{
\text{SGD uses one example; Mini-Batch GD uses a small batch.}
}
$$

## Q4. Learning Rate

### Question

What happens if the learning rate is extremely small or extremely large?

### Solution

### Extremely Small Learning Rate

Updates become very small:

$$
\theta_{\text{new}}\approx\theta_{\text{old}}
$$

Training may be very slow and require many iterations.

### Extremely Large Learning Rate

Updates become very large:

$$
|\eta\nabla L|\text{ is large}
$$

The optimizer may repeatedly overshoot the minimum, causing unstable training or divergence.

### Final Answer

$$
\boxed{
\text{Small }\eta\rightarrow\text{slow training}
}
$$

$$
\boxed{
\text{Large }\eta\rightarrow\text{unstable or possibly divergent training}
}
$$

## Q5. L1 or L2 Regularization

Given:

$$
L_{\text{total}}
=
L+\lambda\sum_iw_i^2
$$

### Step 1 — Identify the Penalty

The weights are squared:

$$
w_i^2
$$

Therefore, this is:

$$
\boxed{\text{L2 Regularization}}
$$

### Step 2 — Purpose

L2 regularization adds a penalty for large weights.

Its goal is to:

- Discourage overly complex models
- Reduce overfitting
- Keep model weights from becoming excessively large

### Final Answer

$$
\boxed{
\text{L2 regularization; it penalizes large weights and helps control overfitting.}
}
$$

## Q6. Complete Optimization Pipeline

### Question

Explain the following optimization pipeline:

$$
\boxed{
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Optimizer}
\rightarrow
\text{Parameter Update}
}
$$

### Solution

### 1. Prediction

The model receives input $X$ and current parameters $\theta$:

$$
\hat{y}=f(X;\theta)
$$

It produces a prediction $\hat{y}$.

### 2. Loss

The prediction is compared with the actual target:

$$
L(y,\hat{y})
$$

The loss measures how wrong the model's prediction is.

### 3. Gradient

Calculate how the loss changes with respect to each parameter:

$$
\nabla_{\theta}L
$$

The gradient tells us the direction in which the loss increases most rapidly.

Therefore, the optimizer can use it to determine how the parameters should change to reduce the loss.

### 4. Optimizer

The optimizer determines how to use the gradient to update the parameters.

Examples:

- Gradient Descent
- SGD
- Momentum
- Adam

### 5. Parameter Update

For basic Gradient Descent:

$$
\theta_{\text{new}}
=
\theta-\eta\nabla_{\theta}L
$$

The parameters are changed so that the next prediction can ideally have a lower loss.

### 6. Repeat

The entire process repeats for many batches and iterations:

$$
\boxed{
\text{Predict}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Update}
\rightarrow
\text{Repeat}
}
$$

### Final Answer

$$
\boxed{
\begin{aligned}
&\text{The model predicts,}\\
&\text{measures its error,}\\
&\text{calculates how each parameter affects that error,}\\
&\text{uses an optimizer to update the parameters,}\\
&\text{and repeats this process to reduce the loss.}
\end{aligned}
}
$$
