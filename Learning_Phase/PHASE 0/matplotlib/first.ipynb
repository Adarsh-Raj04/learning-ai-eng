```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
plt.plot(x, y) # Plotting the data points
plt.show() # Displaying the plot
```

```python
plt.plot(x, y)

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("X vs Y")

plt.show()
```

```python
plt.scatter(x, y) # scatter plot of the data points

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Relationship")

plt.show()
```

```python
# another example of scatter plot with different data points
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "age": [25, 32, 28, 40, 35],
    "salary": [50000, 75000, 62000, 90000, 85000]
})

plt.scatter(df["age"], df["salary"])

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")

plt.show()
```

```python
# Histogram example
plt.hist(df["salary"])

plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")

plt.show()
```

```python
df["department"] = ["AI", "Data", "AI", "Data", "AI"]

avg_salary = df.groupby("department")["salary"].mean()

plt.bar(avg_salary.index, avg_salary.values)

plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.title("Average Salary by Department")

plt.show()
```
