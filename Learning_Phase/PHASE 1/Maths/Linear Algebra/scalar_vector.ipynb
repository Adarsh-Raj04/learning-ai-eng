# Linear Algebra Foundations

## Scaler

- A scaler is simply a number
- We normally represent a scalar using a lowercase letter: x = 5 where x is a variable and 5 is it's value.

**Scaler in Python**
```python
x = 5
y = 2

print(x + y) = 7
print(x - y) = 3
print(x * y) = 10
print(x / y) = 2.5
```
---
**Scaler in Numpy**
```python
import numpy as np

x = np.array(5)

print(x) = 5
print(x.shape) = ()
print(x.ndim) = 0 
```
A scalar has **0 dimensions.**

---
**Scaler in ML:** A scalar is a single numerical quantity. It is the fundamental building block used to represent individual values in mathematical and ML computations.

## Vector

- A vector is an ordered collection of numbers.

Example:
$$
\mathbf{x} =
\begin{bmatrix}
28\\
180\\
72
\end{bmatrix}
$$

This could represent
```text
Age    = 28
Height = 180
Weight = 72
```

**2. Why do vectors exist?**

This is extremely important for ML.

Imagine you're giving information about a house to a model:

```text
Area       = 1500 sq ft
Bedrooms   = 3
Bathrooms  = 2
Age        = 10 years
```

We can represent the house as:

$$
\mathbf{x}=
\begin{bmatrix}
1500\\
3\\
2\\
10
\end{bmatrix}
$$

Now the entire house can be represented as **one feature vector**.

This is exactly the kind of representation ML models work with.

```text
Real-world object
       ↓
Features
       ↓
Vector
       ↓
ML model
```

---
**Mathematical Notations**
We usually represent a vector using a bold lowercase letter: x=[2,5,7]

**Position Matters**
The order matters.

Generally:

$$
[2,5,7] \neq [7,5,2]
$$

---
**Dimension of a Vector**

If a vector contains \(n\) elements, we call it an **n-dimensional vector**.

For example:

$$
\mathbf{x}=[5]
$$

→ 1-dimensional vector.

$$
\mathbf{x}=[5,10,15]
$$

→ 3-dimensional vector.

$$
\mathbf{x}=[2,4,6,8,10]
$$

→ 5-dimensional vector.

So:

$$
\boxed{\text{Number of elements = vector dimension}}
$$

Don't confuse this with `ndim` in NumPy yet. We'll make that distinction very clear when we get there.

**Vector Magnitude**
- A vector has direction and size. Magnitude answers - How large/long is this vector.
- It's formula:
$$
\boxed{
\|\mathbf{x}\|=\sqrt{x_1^2+x_2^2+\cdots+x_n^2}
}
$$
- This is also called the Euclidean norm, or L2 norm.
- Magnitude becomes useful when comparing embeddings, calculating distances, normalizing vectors, and measuring parameter sizes.

````python
import numpy as np

x = np.array([3, 4])

magnitude = np.linalg.norm(x) # under the hood it performs np.sqrt(np.sum(x ** 2))

print(magnitude)
````

**Distance between two vectors**
- Take a = [1,2] and b = [4,6]
- First Find their difference: a-b = [-3,-4]
- Then take it's magnitude:
$$
\|\mathbf{a}-\mathbf{b}\|
=
\sqrt{(-3)^2+(-4)^2}
=
5
$$
- Therefore:

$$
\boxed{d(\mathbf a,\mathbf b)=\|\mathbf a-\mathbf b\|}
$$

### Dot Product
- It's general formula: 
$$
\mathbf a\cdot\mathbf b
=
\sum_{i=1}^{n}a_i b_i
$$
- Dot product has a geometric interpretation:

$$
\boxed{
\mathbf a\cdot\mathbf b
=
\|\mathbf a\|\|\mathbf b\|\cos\theta
}
$$
- So we can calculate: 
$$
\cos\theta =
\frac{\mathbf a\cdot\mathbf b}
{\|\mathbf a\|\|\mathbf b\|}
$$

````python
a = np.array([2, 3])
b = np.array([4, 5])

print(a @ b)
````

### Norms
- A norm is a function that measures the size/magnitude of a vector.
- L1 Norm: 
$$
\boxed{\|x\|_1=|x_1|+|x_2|+\cdots+|x_n|}
$$
- L2 Norm (The most important one for ML):
$$
\|\mathbf{x}\|_2
=
\sqrt{x_1^2+x_2^2+\cdots+x_n^2}
$$
---
- ML connection

Norms are heavily used in **regularization**:

$$
L+\lambda\|W\|
$$

The norm penalizes large model parameters.

### Unit Vector

A **unit vector** is simply a vector whose magnitude is exactly 1.

Suppose:

$$
x=[3,4]
$$

Magnitude:

$$
\|x\|=5
$$

To turn it into a unit vector, divide by its magnitude:

$$
\hat{x}=\frac{x}{\|x\|}
$$

Therefore:

$$
\hat{x}
=
\frac{1}{5}
[3,4]
=
[0.6,0.8]
$$

Check:

$$
\sqrt{0.6^2+0.8^2}=1
$$

So:

$$
\boxed{\|\hat{x}\|=1}
$$

### Normalization

- This process is called **normalization**:

$$
\boxed{x_{\text{normalized}}=\frac{x}{\|x\|}}
$$



- Why embeddings care

    Suppose:

    $$
    A=[3,4]
    $$

    $$
    B=[30,40]
    $$

    They're pointing in exactly the same direction, but their magnitudes differ.

    After normalization:

    $$
    A_n=[0.6,0.8]
    $$

    $$
    B_n=[0.6,0.8]
    $$

    Now they're identical.

    This is one reason normalized embeddings are useful when using **cosine similarity**.

## Projection

This one is worth understanding because it appears later in PCA and linear algebra.

Suppose we have vector \(a\), and we want to find **how much of \(a\) lies in the direction of \(b\)**.

That's a projection.

The scalar projection of \(a\) onto \(b\) is:

$$
\boxed{
\frac{a\cdot b}{\|b\|}
}
$$

If \(b\) is a **unit vector**, this becomes simply:

$$
\boxed{a\cdot b}
$$

The vector projection is:

$$
\boxed{
\operatorname{proj}_b(a)
=
\frac{a\cdot b}{b\cdot b}b
}
$$

### Simple example

Let:

$$
a=[3,4]
$$

and:

$$
b=[1,0]
$$

Here \(b\) is already a unit vector.

Calculate:

$$
a\cdot b=3(1)+4(0)=3
$$

So the projection of \(a\) onto the x-axis is:

$$
\boxed{[3,0]}
$$

Visually:

```text
a = (3,4)
     •
     |\
     | \
     |  \
     |   \
-----•----•------→
     0    3
```

We're essentially asking:

> "If I drop this vector onto the x-axis, what part remains?"

Answer:

$$
[3,0]
$$
