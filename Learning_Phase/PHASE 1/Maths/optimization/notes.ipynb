# Phase 1 — Optimization Complete Notes

## 1. Objective Function / Loss Function

### Definition

An **objective function** is the function we want to optimize.

In ML, we usually minimize a **loss function** that measures how wrong the model's predictions are.

### Formula

For Mean Squared Error (MSE):

$$
L = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Where:

- $n$ = number of training examples
- $y_i$ = actual value
- $\hat{y}_i$ = predicted value

### ML Relevance

Training means finding model parameters that minimize the loss:

$$
\boxed{\text{Find parameters that minimize } L}
$$

---

## 2. Convexity

### Definition

A **convex function** has a bowl-like shape with a global minimum.

### Formula

A function $f(x)$ is convex if:

$$
f(\lambda x_1+(1-\lambda)x_2)
\leq
\lambda f(x_1)+(1-\lambda)f(x_2)
$$

Where:

- $x_1, x_2$ = two points
- $0 \leq \lambda \leq 1$

### ML Relevance

For a convex optimization problem:

$$
\boxed{\text{Local Minimum}=\text{Global Minimum}}
$$

Linear regression with MSE is generally convex, while neural-network optimization is generally non-convex.

---

## 3. Gradient Descent

### Definition

**Gradient Descent** repeatedly changes model parameters in the direction that reduces the loss.

### Formula

$$
\boxed{
\theta_{\text{new}}
=
\theta_{\text{old}}
-
\eta\nabla L(\theta)
}
$$

Where:

- $\theta$ = model parameters
- $L(\theta)$ = loss function
- $\nabla L(\theta)$ = gradient of the loss
- $\eta$ = learning rate

### ML Relevance

Gradient Descent is a fundamental optimization method used to train many ML and deep-learning models.

---

## 4. Stochastic Gradient Descent (SGD)

### Definition

**Stochastic Gradient Descent (SGD)** calculates the gradient using one training example at a time.

### Formula

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta\nabla L_i(\theta_t)
}
$$

Where:

- $\theta_t$ = parameters at step $t$
- $L_i$ = loss for one training example
- $\eta$ = learning rate

### ML Relevance

SGD allows frequent updates and can be useful for large datasets, but its updates are noisier than batch Gradient Descent.

---

## 5. Mini-Batch Gradient Descent

### Definition

**Mini-Batch Gradient Descent** calculates the gradient using a small group of training examples.

### Formula

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta
\frac{1}{B}
\sum_{i=1}^{B}
\nabla L_i(\theta_t)
}
$$

Where:

- $B$ = batch size
- $\eta$ = learning rate
- $L_i$ = loss for example $i$
- $\theta_t$ = parameters at step $t$

### ML Relevance

Mini-batch training is the standard practical approach for training neural networks because it balances computational efficiency and gradient stability.

---

## 6. Momentum

### Definition

**Momentum** uses previous gradient information to make optimization updates smoother and potentially faster.

### Formula

The velocity/update direction is:

$$
v_t
=
\beta v_{t-1}
+
(1-\beta)\nabla L(\theta_t)
$$

The parameter update is:

$$
\theta_{t+1}
=
\theta_t
-
\eta v_t
$$

Where:

- $v_t$ = accumulated update direction
- $\beta$ = momentum factor
- $\eta$ = learning rate
- $\nabla L(\theta_t)$ = current gradient

### ML Relevance

Momentum can reduce zig-zagging and accelerate movement in directions where gradients consistently point in the same direction.

---

## 7. Adam

### Definition

**Adam (Adaptive Moment Estimation)** combines momentum-like gradient estimates with adaptive scaling of parameter updates.

### Step 1 — First Moment Estimate

$$
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t
$$

### Step 2 — Second Moment Estimate

$$
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

### Step 3 — Bias-Corrected Estimates

$$
\hat{m}_t
=
\frac{m_t}{1-\beta_1^t}
$$

$$
\hat{v}_t
=
\frac{v_t}{1-\beta_2^t}
$$

### Step 4 — Parameter Update

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta
\frac{\hat{m}_t}
{\sqrt{\hat{v}_t}+\epsilon}
}
$$

Where:

- $g_t$ = current gradient
- $m_t$ = first-moment estimate
- $v_t$ = second-moment estimate
- $\hat{m}_t$ = bias-corrected first moment
- $\hat{v}_t$ = bias-corrected second moment
- $\beta_1, \beta_2$ = decay rates
- $\eta$ = learning rate
- $\epsilon$ = small constant that improves numerical stability

### ML Relevance

Adam is widely used for neural-network training because it adapts the effective update scale for different parameters while incorporating momentum-like information.

---

## 8. Learning Rate

### Definition

The **learning rate** controls the size of each parameter update.

### Formula

$$
\theta_{\text{new}}
=
\theta
-
\eta\nabla L
$$

Where $\eta$ is the learning rate.

### ML Relevance

#### Learning Rate Too Small

- Training can be very slow.
- Many iterations may be required.

#### Learning Rate Too Large

- Updates can overshoot the minimum.
- Training may become unstable.
- The optimization process may diverge.

---

## 9. Regularization

### Definition

**Regularization** adds a penalty to the loss function to discourage overly complex models and reduce overfitting.

### L1 Regularization

$$
L_{\text{total}}
=
L_{\text{original}}
+
\lambda\sum_i|w_i|
$$

Where:

- $w_i$ = model weights
- $\lambda$ = regularization strength

L1 regularization tends to push some weights toward exactly zero.

### L2 Regularization

$$
L_{\text{total}}
=
L_{\text{original}}
+
\lambda\sum_iw_i^2
$$

L2 regularization discourages very large weights.

### ML Relevance

Regularization helps control model complexity and can improve generalization.

---

## 10. Complete ML Optimization Loop

### Definition

Model training repeatedly performs:

1. Prediction
2. Loss calculation
3. Gradient calculation
4. Optimization
5. Parameter update

### Flow

$$
\boxed{
\text{Input}
\rightarrow
\text{Model}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Gradient}
\rightarrow
\text{Optimizer}
\rightarrow
\text{Parameter Update}
\rightarrow
\text{Repeat}
}
$$

### Mathematical View

#### Prediction

$$
\hat{y}=f(X;\theta)
$$

#### Loss

$$
L(y,\hat{y})
$$

#### Gradient

$$
\nabla_{\theta}L
$$

#### Parameter Update

For a generic optimizer:

$$
\theta_{\text{new}}
=
\theta
-
\text{Optimizer Step}
$$

For basic Gradient Descent:

$$
\theta_{\text{new}}
=
\theta
-
\eta\nabla_{\theta}L
$$

### ML Relevance

This is the core training loop behind linear models, neural networks, deep learning, and large-scale model training.

| Momentum | Use previous gradient information |
