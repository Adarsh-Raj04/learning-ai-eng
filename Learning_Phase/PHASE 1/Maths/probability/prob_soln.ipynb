# Phase 1 — Probability Practice Problems & Solutions

## Q1. Conditional Probability

Given:

$$
P(A)=0.5,\quad P(B)=0.4,\quad P(A\cap B)=0.2
$$

Find $P(A|B)$.

### Solution

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

$$
=\frac{0.2}{0.4}
$$

$$
\boxed{P(A|B)=0.5}
$$

### Check Independence

For independent events:

$$
P(A|B)=P(A)
$$

Here:

$$
P(A|B)=0.5
$$

and:

$$
P(A)=0.5
$$

Therefore:

$$
\boxed{\text{A and B are independent}}
$$

## Q2. Bayes' Theorem

Given:

$$
P(A)=0.2,\quad P(B|A)=0.8,\quad P(B)=0.5
$$

### Solution

Using Bayes' theorem:

$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$

Substitute the values:

$$
=\frac{(0.8)(0.2)}{0.5}
$$

$$
=\frac{0.16}{0.5}
$$

$$
\boxed{P(A|B)=0.32}
$$

## Q3. Expectation

Given:

| $X$ | Probability |
| --: | ----------: |
|   0 |         0.2 |
|   1 |         0.5 |
|   2 |         0.3 |

### Solution

The expectation is:

$$
E[X]=\sum_x xP(X=x)
$$

Therefore:

$$
E[X]=(0)(0.2)+(1)(0.5)+(2)(0.3)
$$

$$
=0+0.5+0.6
$$

$$
\boxed{E[X]=1.1}
$$

---

## Q4. Variance

Use the same distribution from Q3.

### Solution

From Q3:

$$
E[X]=1.1
$$

First calculate $E[X^2]$:

$$
E[X^2]=(0^2)(0.2)+(1^2)(0.5)+(2^2)(0.3)
$$

$$
=0+0.5+1.2
$$

$$
E[X^2]=1.7
$$

Now use:

$$
Var(X)=E[X^2]-(E[X])^2
$$

Therefore:

$$
Var(X)=1.7-(1.1)^2
$$

$$
=1.7-1.21
$$

$$
\boxed{Var(X)=0.49}
$$

## Q5. Covariance

Given:

$$
X=[1,2,3]
$$

$$
Y=[2,4,6]
$$

### Solution

As $X$ increases, $Y$ also increases.

In fact:

$$
Y=2X
$$

Therefore, $X$ and $Y$ move together.

Hence:

$$
\boxed{\text{Covariance is positive}}
$$

For reference, the exact covariance is:

$$
Cov(X,Y)=\frac{4}{3}
$$

But the question only asks for the direction.

## Q6. Bernoulli

Given:

$$
P(\text{spam})=0.7
$$

and $X=1$ means spam.

### Solution

Since:

$$
P(X=1)=0.7
$$

The probability of $X=0$ is:

$$
P(X=0)=1-p
$$

Therefore:

$$
P(X=0)=1-0.7
$$

$$
\boxed{P(X=0)=0.3}
$$

## Q7. Binomial

Given:

$$
n=3,\quad k=2,\quad p=0.8
$$

We want the probability that exactly 2 samples are classified correctly.

### Solution

The binomial probability formula is:

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$

Substitute the values:

$$
P(X=2)=\binom{3}{2}(0.8)^2(0.2)^1
$$

Calculate the combination:

$$
\binom{3}{2}=3
$$

Therefore:

$$
P(X=2)=3(0.64)(0.2)
$$

$$
=3(0.128)
$$

$$
\boxed{P(X=2)=0.384}
$$

As a percentage:

$$
\boxed{38.4\%}
$$

## Q8. Gaussian

Given:

$$
X\sim N(100,15^2)
$$

### Solution

Compare this with the standard notation:

$$
X\sim N(\mu,\sigma^2)
$$

Therefore:

$$
\boxed{\mu=100}
$$

$$

## Q9. Likelihood

Observed sequence:

```text
Heads, Heads, Tails, Heads
```

Let:

$$
p=P(\text{Heads})
$$

Therefore:

$$
P(\text{Tails})=1-p
$$

### Solution

There are 3 Heads and 1 Tail.

The probability of observing this exact sequence is:

$$
L(p)=p\cdot p\cdot(1-p)\cdot p
$$

Therefore:

$$
\boxed{L(p)=p^3(1-p)}
$$

## Q10. MLE

Observed sequence:

```text
Heads, Heads, Tails, Heads
```

### Solution

For a Bernoulli distribution, the MLE of $p$ is the observed proportion of successes.

$$
\hat p_{MLE}
=
\frac{\text{Number of Heads}}
{\text{Total Observations}}
$$

There are 3 Heads out of 4 observations:

$$
\hat p_{MLE}=\frac{3}{4}
$$

Therefore:

$$
\boxed{\hat p_{MLE}=0.75}
$$

## Q11. MLE vs MAP

### Solution

### Maximum Likelihood Estimation

MLE chooses the parameter that maximizes the likelihood of the observed data:

$$
\boxed{
\theta_{MLE}
=
\arg\max_{\theta}P(Data|\theta)
}
$$

### Maximum A Posteriori

MAP considers both the likelihood and a prior:

$$
\boxed{
\theta_{MAP}
=
\arg\max_{\theta}
P(Data|\theta)P(\theta)
}
$$

Therefore:

$$
\boxed{
\text{MLE = likelihood}
}
$$

$$
\boxed{
\text{MAP = likelihood + prior}
}
$$

The key difference is that **MLE does not include a prior**, while **MAP does**.

## Q12. LLM Token Probabilities

Given:

```text
P("AI" | "I am learning")      = 0.60
P("ML" | "I am learning")      = 0.25
P("Python" | "I am learning")  = 0.15
```

### Solution

These probabilities represent a probability distribution over possible next tokens.

```text
AI       → 60%
ML       → 25%
Python   → 15%
```

Check that the probabilities sum to 1:

$$
0.60+0.25+0.15=1
$$

Therefore, this is a valid probability distribution over these possible outcomes.

The meaning of:

$$
P(\text{"AI"}|\text{"I am learning"})=0.60
$$

is:

> Given the context `"I am learning"`, the model assigns a 60% probability to `"AI"` being the next token.

It does **not** mean that there is a 60% chance that the person is learning AI.
