```python
# check for missing values

df.isnull()  # check for missing values in the dataframe, we can use df.isna() as well

df.isna().sum() # count of missing values in each column
```

```python
# remove rows with missing values
modified_df = df.dropna()  # remove rows with missing values
```

```python
# verify the changes
modified_df.info()  # display information about the modified dataframe
```

```python
# change the data type of the age and salary columns to integer
modified_df['age'] = modified_df['age'].astype('Int64')  # change the data type of the age column to integer
modified_df['salary'] = modified_df['salary'].astype('Int64')  # change the data type of the salary column to integer
```

```python
mod_df = df.fillna({'age': 0, 'salary': 0})  # fill missing values in the age and salary column with 0

#change the data type of the age and salary columns to integer
mod_df['age'] = mod_df['age'].astype('Int64')  # change the data type of the age column to integer
mod_df['salary'] = mod_df['salary'].astype('Int64')  # change the data type of the salary column to integer



# we can also fill missing values with a specific value, for example, we can fill missing values in the age column with the mean age and in the salary column with the median salary
mean_age = mod_df['age'].mean()  # calculate the mean age
median_salary = mod_df['salary'].median()  # calculate the median salary

print(f"Mean age: {mean_age}, Median salary: {median_salary}")  # print the mean age and median salary

mod_df = mod_df.fillna({'age': mean_age, 'salary': median_salary})  # fill missing values in the age column with the mean age and in the salary column with the median salary

```

```python
# check the info of the dataframe
mod_df.info()  # display information about the dataframe
```

```python
# sorting by one column
sorted_df = mod_df.sort_values(by='age')  # sort the dataframe by the age column
    

#get unique values of age column with total number of each unique values
unique_ages = sorted_df['age'].value_counts()
print(f"Unique ages: {unique_ages}")
```

```python
# sortign by multiple columns
df1 = df.sort_values(by=['department', 'salary'], ascending=[True, False])  # sort the dataframe by the department column in ascending order and by the salary column in descending order

df1.head()  # display the first few rows of the dataframe after sorting
```

```python
# grouping by one column
print(df.groupby("department")["salary"].mean()) # group the dataframe by the department column and calculate the mean salary for each department


print(df.groupby("department")["salary"].agg(
    ["mean", "min", "max"]
))  # group the dataframe by the department column and calculate the mean, min, and max salary for each department


print(df.groupby("department").agg({
    "salary": "mean",
    "age": "mean"
}))  # group the dataframe by the department column and calculate the mean salary and mean age for each department
```

```python
# use of apply

def salary_category(salary):
    if salary >= 80000:
        return "High"
    elif salary >= 60000:
        return "Medium"
    else:
        return "Low"

df["salary_category"] = df["salary"].apply(salary_category)

print(df)  # display the dataframe with the new salary_category column


# we can use lambda function to achieve the same result as above
df["salary_category_lambda"] = df["salary"].apply(lambda x: "High" if x >= 80000 else "Medium" if x >= 60000 else "Low")

print(df)  # display the dataframe with the new salary_category_lambda column
```

## Merge and Join

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3],
    "name": ["Alice", "Bob", "Charlie"]
})

details = pd.DataFrame({
    "employee_id": [1, 2, 3],
    "department": ["AI", "Data", "AI"],
    "salary": [50000, 75000, 62000]
})
```

```python
# merge the two dataframes on the employee_id column
result = employees.merge(
    details,
    on="employee_id"
)

print(result)  # display the merged dataframe
```

```python
# Join Types 
## Inner Join: Returns records that have matching values in both tables.
inner_join = employees.merge(details, on="employee_id", how="inner")
print(f"Inner Join Result:\n{inner_join}")

## Left Join: Returns all records from the left table (employees), and the matched records from the right table (details). If there is no match, the result is NULL on the right side.
left_join = employees.merge(details, on="employee_id", how="left")
print(f"Left Join Result:\n{left_join}")

## Right Join: Returns all records from the right table (details), and the matched records from the left table (employees). If there is no match, the result is NULL on the left side.
right_join = employees.merge(details, on="employee_id", how="right")
print(f"Right Join Result:\n{right_join}")

## outer Join: Returns all records when there is a match in either left (employees) or right (details) table records. If there is no match, the result is NULL from the side that does not have a match.
outer_join = employees.merge(details, on="employee_id", how="outer")
print(f"Outer Join Result:\n{outer_join}")
```
