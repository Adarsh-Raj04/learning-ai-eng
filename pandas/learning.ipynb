```python
# Series

import pandas as pd

s = pd.Series([10, 20, 30, 40])

print(s)

s = pd.Series(
    [10, 20, 30],
    index=["A", "B", "C"]
)

print(s) # with custom index
```

```python
# Creating DataFrame
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "salary": [50000, 60000, 80000]
})
```

```python
# Inspecting DataFrame

print(f"First 5 rows:\n{df.head()}") # first 5 rows [default set to 5]

print(f"\nFirst 2 rows:\n{df.head(2)}") # first 2 rows

print(f"\nLast 5 rows:\n{df.tail()}") # last 5 rows [default set to 5]

print(f"\nShape of the DataFrame: {df.shape}") # shape of the DataFrame

print(f"\nColumn names: {df.columns}") # column names

print(f"\nData types:\n{df.dtypes}") # data types of each column

print(f"\nSummary: \n{df.info()}") # summary of the DataFrame

print(f"\nStatistical summary: \n{df.describe()}") # statistical summary of the DataFrame
```

```python
# Selecting Columns and Rows

print(f"\nSelected column:\n{df['name']}") # select a single column

print(f"\nSelected columns:\n{df[['name', 'age']]}") # select multiple columns

print(f"\nSelected row:\n{df.iloc[1]}") # select a single row by index

print(f"\nSelected rows:\n{df.iloc[1:3]}") # select multiple rows by index


# label-based selection using loc

print(f"\nSelected row by label:\n{df.loc[1]}") # select a single row by label

print(f"\nSelected rows by label:\n{df.loc[1:2]}") # select multiple rows by label
```

```python
# filtering rows based on conditions

df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [25, 30, 35, 22],
    "salary": [50000, 60000, 80000, 45000]
})

print(f"\nRows where age > 25:\n{df[df['age'] > 25]}") # filter rows where age is greater than 25

print(f"\nRows where salary > 60000:\n{df[df['salary'] >= 60000]}") # filter rows where salary is greater than 60000

print(f"\nRows where age > 25 and salary > 60000:\n{df[(df['age'] > 25) & (df['salary'] > 60000)]}") # filter rows where age is greater than 25 and salary is greater than 60000
```

```python
# creating new columns
df['bonus'] = df['salary'] * 0.1 # create a new column 'bonus' which is 10% of salary

df["age_plus_5"] = df["age"] + 5

print(f"\nDataFrame with new column:\n{df}") # print the DataFrame with the new column
```

```python
# Basic Operations
print(f"\nMean salary: {df['salary'].mean()}")
print(f"\nMax salary: {df['salary'].max()}")
print(f"\nMin salary: {df['salary'].min()}")
print(f"\nSum salary: {df['salary'].sum()}")
print(f"\nStandard deviation of salary: {df['salary'].std()}")


# get a quick statistical summary of the DataFrame
print(f"\nStatistical summary:\n{df.describe()}") # statistical summary of the DataFrame
```
